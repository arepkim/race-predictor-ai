```python
import pandas as pd
import numpy as np

# Set seed for reproducibility
np.random.seed(42)
num_runners = 1000

# 1. Identity & Physiological Baselines
runner_ids = [f"R-{str(i).zfill(4)}" for i in range(1, num_runners + 1)]
ages = np.random.randint(18, 60, size=num_runners)
gender = np.random.choice(['Male', 'Female'], size=num_runners)

# Capturing extreme biological variance 
resting_hr = np.random.randint(45, 76, size=num_runners) 
max_hr = np.random.randint(170, 209, size=num_runners) 
vo2_max = np.round(np.random.uniform(35.0, 65.0, size=num_runners), 1)

# 2. The 12-Week Training Block
performance_factor = (vo2_max - 35) / 30  # Scale 0 to 1

weekly_mileage = np.round(20 + (performance_factor * 80) + np.random.normal(0, 10, num_runners), 1)
avg_easy_pace = np.round(8.0 - (performance_factor * 3.0) + np.random.normal(0, 0.5, num_runners), 2)
avg_tempo_pace = np.round(avg_easy_pace * 0.85, 2)
training_consistency = np.round(np.random.uniform(0.6, 1.0, size=num_runners), 2)

# 3. Historical PBs (Anchored to their VO2 max and pace)
pb_5k_mins = np.round((avg_tempo_pace * 5) * (1.05 - (performance_factor * 0.1)), 1)
pb_10k_mins = np.round(pb_5k_mins * 2.1, 1)
pb_half_mins = np.round(pb_10k_mins * 2.22, 1)
pb_full_mins = np.round(pb_half_mins * 2.15, 1)

# 4. Target Variables
races = ['5K', '10K', 'Half Marathon', 'Full Marathon']
target_race = np.random.choice(races, size=num_runners)
actual_finish_time = []

# Generate actual finish times with Gaussian noise
for i in range(num_runners):
    race = target_race[i]
    if race == '5K':
        base_time = pb_5k_mins[i]
    elif race == '10K':
        base_time = pb_10k_mins[i]
    elif race == 'Half Marathon':
        base_time = pb_half_mins[i]
    else:
        base_time = pb_full_mins[i]
    
    # Add random day-of-race variance (e.g., weather, fatigue)
    variance = np.random.normal(0, base_time * 0.03) 
    actual_finish_time.append(np.round(base_time + variance, 1))

# 5. Compile and Export
df = pd.DataFrame({
    'runner_id': runner_ids,
    'age': ages,
    'gender': gender,
    'resting_hr': resting_hr,
    'max_hr': max_hr,
    'vo2_max': vo2_max,
    'weekly_mileage_km': np.clip(weekly_mileage, 15, 140),
    'avg_easy_pace_min_km': np.clip(avg_easy_pace, 4.0, 9.0),
    'avg_tempo_pace_min_km': np.clip(avg_tempo_pace, 3.5, 7.5),
    'training_consistency': training_consistency,
    'pb_5k_mins': pb_5k_mins,
    'pb_10k_mins': pb_10k_mins,
    'pb_half_mins': pb_half_mins,
    'pb_full_mins': pb_full_mins,
    'target_race': target_race,
    'actual_finish_time_mins': actual_finish_time
})

# Save to CSV
df.to_csv('fyp_runner_dataset.csv', index=False)
print("Dataset successfully generated and saved as 'fyp_runner_dataset.csv'!")
```
